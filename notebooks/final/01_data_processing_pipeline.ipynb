# 01. 최종 데이터 전처리 파이프라인

최종 제출 기준 데이터셋을 생성한다. 전처리 로직은 `pipeline/data_processing.py`를 단일 기준으로 사용하며, target은 영업일 기준 5일 초과 지연 여부로 정의한다.

```python
from pathlib import Path
import sys


def find_project_root() -> Path:
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / 'pipeline' / 'data_processing.py').exists():
            return candidate
    raise FileNotFoundError('프로젝트 루트를 찾을 수 없습니다.')


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

PROJECT_ROOT
```

```python
from pipeline.data_processing import make_paths, write_fixed_datasets


paths = make_paths(data_dir=PROJECT_ROOT / 'data', output_dir=PROJECT_ROOT / 'data')
train_fixed, test_fixed = write_fixed_datasets(paths)
```

```python
summary = {
    'train_shape': train_fixed.shape,
    'test_shape': test_fixed.shape,
    'train_target_rate': train_fixed['target'].mean(),
    'test_target_rate': test_fixed['target'].mean(),
}
summary
```
